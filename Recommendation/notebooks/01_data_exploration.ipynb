```python
def explore_dataset(file_path, dataset_name):
    print("=" * 80)
    print(dataset_name)
    print("=" * 80)

    df = pd.read_excel(file_path)

    print("\nShape:")
    print(df.shape)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nFirst 5 rows:")
    display(df.head())

    print("\nData types:")
    print(df.dtypes)

    print("\nMissing values:")
    print(df.isnull().sum())

    print("\nDuplicate rows:")
    print(df.duplicated().sum())

    return df
```

# Overall Dataset Mapping

The seven datasets provide complementary information for personalized learning-path recommendation.

| Dataset | Main Information | Key |
|---|---|---|
| Student Information | Student profile | Student_ID |
| Curriculum & Lesson | Available learning content | Lesson_ID |
| Activity Performance | Current activity performance | Student_ID, Lesson_ID |
| Assessment Results | Academic, motor, creativity and engagement performance | Student_ID |
| Teacher Observation | Teacher-assessed learner characteristics | Student_ID |
| Historical Learning Path | Previous learning sequence and outcomes | Student_ID, Lesson_ID |
| AI Learning Path | Learning progression and intervention rules | ID |

## Overall Data Flow

Student Profile
↓
Activity Performance + Assessment
↓
Teacher Observation
↓
Historical Learning Path
↓
Curriculum and Learning-Path Rules
↓
Personalized Learning Recommendation
